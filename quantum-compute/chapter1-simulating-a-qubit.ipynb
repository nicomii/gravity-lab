```python
# in pennylane every quantum computation runs on a device object
# default qubit and 1 qubit (wires)
device = qml.device(name="default.qubit", wires=1)
# qnode = quantum function + quantum device simulator

@qml.qnode(device=device)
def qubit():
    #literallt a qubit is that easy
    state = qml.state()
    return state
```

```python
print(qubit())
```

that's the 0 state qubit, to sim the 1 state, we run it thorough a x gate (aka pauli x gate/not gate)

```python
@qml.qnode(device=device)
def qubit():
    #literallt a qubit is that easy
    # this time we put it through pennylane's x function
    state = qml.X(wires=0)
    state = qml.state()
    return state
```

```python
print(qubit())
```
